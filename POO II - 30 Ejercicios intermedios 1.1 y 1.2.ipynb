### Ejercicio 1 · Factura con descuento e IVA ★★

Pida el nombre de un producto, la cantidad y el precio unitario. Calcule:
- Subtotal = cantidad × precio
- Descuento del 10 % sobre el subtotal
- IVA del 15 % sobre (subtotal − descuento)
- Total

Muestre la factura con los valores **alineados a la derecha** en 10 espacios y 2 decimales:
```
Producto: Teclado
Subtotal:      90.00
Descuento:      9.00
IVA:           12.15
TOTAL:         93.15
```

````python
producto = input("producto:")
cantidad = int(input("cantidad:"))
precio = float(input("precio:"))

subtotal = cantidad * precio
descuento = subtotal * 0.10
iva = (subtotal - descuento) * 0.15
total = subtotal - descuento + iva

print(f"Producto: {producto}")
print(f"Subtotal:  {subtotal:>10.2f}")
print(f"Descuento: {descuento:>10.2f}")
print(f"IVA:       {iva:>10.2f}")
print(f"TOTAL:     {total:>10.2f}")
````

### Ejercicio 14 · Dígitos de un número ★★

Pida un número entero de 4 cifras (por ejemplo `4827`). Usando solo `//` y `%`:
1. Separe sus 4 dígitos.
2. Muestre la suma de los dígitos.
3. Construya el número invertido como `int` (7284) y súmele 1 para comprobar que es un número.

````python
# 1. Pedir el número entero de 4 cifras
numero = int(input("Ingrese un número entero de 4 cifras: "))

# Separar los 4 dígitos
d1 = numero // 1000
d2 = (numero // 100) % 10
d3 = (numero // 10) % 10
d4 = numero % 10

print(f"Dígitos separados: {d1}, {d2}, {d3}, {d4}")

# 2. Mostrar la suma de los dígitos
suma_digitos = d1 + d2 + d3 + d4
print("La suma de los dígitos es:", suma_digitos)

# 3. Construir el número invertido y sumarle 1
# Para invertirlo (ej. 4827 -> 7284), multiplicamos cada dígito por su posición
numero_invertido = (d4 * 1000) + (d3 * 100) + (d2 * 10) + d1
comprobacion = numero_invertido + 1

print("Número invertido:", numero_invertido)
print("Número invertido + 1:", comprobacion)
````

### Ejercicio 16 · Limpiar un registro ★★

Un sistema exportó este registro con espacios y mayúsculas desordenadas:
`registro = "  0104567896 ,  pérez  ANA ,  ANA.PEREZ@MAIL.COM ,  150.50  "`
1. Sepárelo por comas con `split`.
2. Limpie cada parte con `strip`.
3. Arme un diccionario con `cedula`, `nombre` (formato título y un solo espacio entre palabras), `correo` (minúsculas) y `saldo` (`float`).
4. Muéstrelo.

💡 Para dejar un solo espacio: `" ".join(texto.split())`.

````python
registro = "  0104567896 ,  pérez  ANA ,  ANA.PEREZ@MAIL.COM ,  150.50  "

cedula, nombre, correo, saldo = registro.split(",")
cedula = cedula.strip()
nombre = " ".join(nombre.split()).title()
correo = correo.strip().lower()
saldo = float(saldo.strip())
registro = {
    "cedula": cedula,
    "nombre": nombre,
    "correo": correo,
    "saldo": saldo
}
print(registro)
````

### Ejercicio 21 · Lista de calificaciones ★★

Con `notas = [8.5, 10, 6.0, 9.25, 10, 7.5]`:
1. Agregue `9.0` al final e inserte `5.5` al inicio.
2. Elimine la nota `6.0` con `remove` y la última con `pop` (muestre cuál eliminó).
3. Muestre el promedio con 2 decimales.
4. Muestre cuántos 10 hay y en qué posición está la primera nota 10.
5. Muestre las 3 mejores notas (ordene de mayor a menor y use slicing).

````python
notas = [8.5, 10, 6.0, 9.25, 10, 7.5]

# 1. Agregue 9.0 al final e inserte 5.5 al inicio
notas.append(9.0)
notas.insert(0, 5.5)
print("1. Lista modificada:", notas)

# 2. Elimine la nota 6.0 con remove y la última con pop 
notas.remove(6.0)
nota_eliminada = notas.pop()
print(f"2. Nota eliminada con pop: {nota_eliminada} | Lista actual:", notas)

# 3. Muestre el promedio con 2 decimales
promedio = sum(notas) / len(notas)
print(f"3. Promedio: {promedio:.2f}")

# 4. Muestre cuántos 10 hay y en qué posición está la primera nota 10
cantidad_dieces = notas.count(10)
primera_posicion_diez = notas.index(10)
print(f"4. Cantidad de 10s: {cantidad_dieces} | Primera posición del 10: {primera_posicion_diez}")

# 5. Muestre las 3 mejores notas
notas_ordenadas = sorted(notas, reverse=True)
tres_mejores = notas_ordenadas[:3]
print("5. Las 3 mejores notas:", tres_mejores)
````

original = {"nombre": "Ana", "saldo": 100}

1. Muestre `original`, `alias` y `copia`.

````python
# Diccionario original y asignaciones
original = {"nombre": "Ana", "saldo": 100}
alias = original          # alias apunta exactamente al mismo objeto en memoria
copia = original.copy()   # copia crea un nuevo objeto independiente con los mismos datos

# Modificaciones que se indican en el ejercicio
alias["saldo"] = 500      # Al modificar alias, como comparte referencia con original, afecta a ambos
copia["saldo"] = 999      # Al modificar copia, como es independiente, solo afecta a copia

# 1. Muestre original, alias y copia.
print("1. Resultados de los diccionarios:")
print("original:", original)
print("alias:", alias)
print("copia:", copia)
# Explicación: 
# 'original' y 'alias' muestran saldo 500 porque apuntan al mismo espacio de memoria.
# 'copia' muestra saldo 999 porque es un diccionario independiente creado con .copy().


# 2. Compruebe con 'is' cuáles son el mismo objeto.
print("\n2. Comprobación con 'is':")
print("¿original es alias?:", original is alias)
print("¿original es copia?:", original is copia)
# Explicación:
# Devuelve True para 'original is alias' porque ambos son el mismo objeto exacto en memoria.
# Devuelve False para 'original is copia' porque .copy() crea un objeto nuevo e independiente.


# 3. Compare con id() las direcciones de memoria.
print("\n3. Direcciones de memoria con id():")
print("id de original:", id(original))
print("id de alias:", id(alias))
print("id de copia:", id(id(original) if False else id(copia))) # Forma limpia abajo
print("id de copia:", id(copia))
# Explicación:
# Los id() de 'original' y 'alias' son exactamente el mismo número, confirmando que comparten la 
# misma referencia. El id() de 'copia' es un número diferente, confirmando que vive en otra dirección.
````

### Ejercicio 30 · Ficha de un socio a partir de texto ★★

Un archivo antiguo guarda cada socio en una sola línea:
`linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"`
1. Separe la línea por `"|"`.
2. Cree un diccionario con `nombre` (formato título), `cedula` y `cuentas`, que a su vez es un diccionario `{tipo: saldo}` (separe cada cuenta por `"="` y convierta el saldo a `float`).
3. Muestre la ficha con este formato:
```
Socio: Ana Pérez (0104567896)
Tipos de cuenta: ahorros, plazo fijo
Saldo total: $6,520.75
```

````python
linea = "ana pérez|0104567896|ahorros=1520.75|plazo fijo=5000"

partes = linea.split("|")
tipo1, saldo1 = partes[2].split("=")
tipo2, saldo2 = partes[3].split("=")
diccionario = {
    "nombre": partes[0].title(),
    "cedula": partes[1], 
    "cuentas": {
        tipo1: float(saldo1),
        tipo2: float(saldo2)
    }
}

saldo_total = sum(diccionario["cuentas"].values())
tipos_cuenta = ", ".join(diccionario["cuentas"].keys())

print(f"Socio: {diccionario['nombre']} ({diccionario['cedula']})")
print(f"Tipos de cuenta: {tipos_cuenta}")
print(f"Saldo total: ${saldo_total:,.2f}")
````
